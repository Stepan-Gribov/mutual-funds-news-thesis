prime

```python
folder_path = r'../../Data/Processed_data/1prime_news_database/1prime_downloaded_news_processed'
# List CSV files in the folder
csv_files = [os.path.join(folder_path, f) for f in os.listdir(folder_path) if f.lower().endswith('.csv')]

dfs = []
for file in csv_files:
    df = pd.read_csv(file,encoding='utf-16', sep= '^')  # Adjust encoding or separator if needed
    # Insert the new column 'source' with a fixed value "interfax" as the first column
    df.insert(0, "news_source", "prime")
    dfs.append(df)
    # print(file)

prime_df = pd.concat(dfs, ignore_index=True)
prime_df.drop(columns=['source_file','source', 'key'], inplace=True)
```

```python
prime_df['date'] = prime_df['date'].apply(
    lambda s: s.split(',')[0].strip() if isinstance(s, str) and ',' in s else s
)

prime_df['date'] = prime_df['date'].apply(lambda s: f"{s} 2025" if isinstance(s, str) and len(s.split()) == 2 else s)

# Словарь для сопоставления русских названий месяцев в родительном падеже с номером месяца
month_map = {
    "января": "01",
    "февраля": "02",
    "марта": "03",
    "апреля": "04",
    "мая": "05",
    "июня": "06",
    "июля": "07",
    "августа": "08",
    "сентября": "09",
    "октября": "10",
    "ноября": "11",
    "декабря": "12"
}

def convert_date_simple(date_str):
    try:
        # Ожидаемый формат: "9 января 2024"
        day, mon, year = date_str.split()
        month_num = month_map.get(mon.lower(), "00")
        return f"{year}-{month_num}-{int(day):02d}"
    except Exception:
        return date_str

# Преобразуем столбец 'date' к формату yyyy-mm-dd
prime_df['date'] = prime_df['date'].apply(convert_date_simple)
prime_df = prime_df.drop(3066)
prime_df = prime_df.sort_values(by=['date'], ascending=[False])
```

```python
rbc_news = pd.read_excel(r'../../Data/Processed_data/missed_news/missed_news_rbc.xlsx')

RBC_df = rbc_news
# RBC_df.drop(columns=['source_file'], inplace=True)
RBC_df = RBC_df.rename(columns={'category': 'date'})
```

```python
RBC_df['date'] = RBC_df['date'].apply(
    lambda s: s.split(',')[-2].strip() if isinstance(s, str) and len(s.split(',')) >= 2 else s
)
```

```python
RBC_df['date'] = RBC_df['date'].apply(lambda s: f"{s} 2025" if isinstance(s, str) and len(s.split()) == 2 else s)

month_map = {
    "янв": "01", "фев": "02", "мар": "03", "апр": "04", "мая": "05", 
    "июн": "06", "июл": "07", "авг": "08", "сен": "09", "окт": "10", 
    "ноя": "11", "дек": "12"
}

def convert_date_simple(date_str):
    try:
        day, mon, year = date_str.split()
        mon = mon.lower()  # Приводим месяц к нижнему регистру
        month_num = month_map.get(mon, "00")
        return f"{year}-{month_num}-{int(day):02d}"
    except Exception:
        return date_str

RBC_df['date'] = RBC_df['date'].apply(convert_date_simple)
RBC_df = RBC_df.sort_values(by=['name','date'], ascending=[True,False])
```

```python
import pandas as pd 
Y_funds_df = pd.read_excel(r'../../Data/Y/Y.xlsx')
```

```python
target_days = Y_funds_df['date'].unique()
```

```python
target_days
```

```python
ETF_flows = pd.read_excel(r'../../Data/Controls/ETF_flows.xlsx')
```

```python
ETF_flows["date_closest"] = ETF_flows["date"].apply(
    lambda x: min(
        [(abs(d - x), d) for d in target_days if abs(d - x) <= pd.Timedelta(days=5)],
        key=lambda t: t[0]
    )[1] if any(abs(d - x) <= pd.Timedelta(days=5) for d in target_days) else pd.NaT
)
ETF_flows["date_closest"].fillna(method="ffill", inplace=True)
```

```python
ETF_flows.to_excel(r'../../Data/Controls/ETF_flows.xlsx', index=False)
```

```python
imoex = pd.read_excel(r'../../Data/Controls/imoex.xlsx')
```

```python
imoex["TRADEDATE"] = pd.to_datetime(imoex["TRADEDATE"])
imoex.sort_values(by=['TRADEDATE'], ascending=[False], inplace=True)

imoex["date_closest"] = imoex["TRADEDATE"].apply(
    lambda x: min(
        [(abs(d - x), d) for d in target_days if abs(d - x) <= pd.Timedelta(days=20)],
        key=lambda t: t[0]
    )[1] if any(abs(d - x) <= pd.Timedelta(days=20) for d in target_days) else pd.NaT
)
# imoex["date_closest"].fillna(method="ffill", inplace=True)
imoex = imoex[imoex["TRADEDATE"] == imoex["date_closest"]]
```

```python
imoex.to_excel(r'../../Data/Controls/imoex.xlsx', index=False)
```

```python
ruonia = pd.read_excel(r'../../Data/Controls/ruonia.xlsx')
```

```python
ruonia["DT"] = pd.to_datetime(ruonia["DT"])
ruonia.sort_values(by=['DT'], ascending=[False], inplace=True)

ruonia["date_closest"] = ruonia["DT"].apply(
    lambda x: min(
        [(abs(d - x), d) for d in target_days if abs(d - x) <= pd.Timedelta(days=20)],
        key=lambda t: t[0]
    )[1] if any(abs(d - x) <= pd.Timedelta(days=20) for d in target_days) else pd.NaT
)
# ruonia["date_closest"].fillna(method="ffill", inplace=True)
ruonia = ruonia[ruonia["DT"] == ruonia["date_closest"]]
```

```python
ruonia.to_excel(r'../../Data/Controls/ruonia.xlsx', index=False)
```
