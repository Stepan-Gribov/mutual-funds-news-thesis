```python
import pandas as pd
```

```python
shares_df = pd.read_csv(r'../../Data/quotes_data/Shares.csv', encoding = 'utf-16', sep = '\t')
```

```python
# ensure sorted by ISIN and date
shares_df = shares_df.sort_values(['ISIN', 'TRADEDATE'])

# compute daily returns
shares_df['ret'] = shares_df.groupby('ISIN')['LEGALCLOSEPRICE'].pct_change()

# 30-day rolling volatility of returns
shares_df['volatility'] = (
    shares_df
    .groupby('ISIN')['ret']
    .rolling(30, min_periods=30)
    .std()
    .reset_index(level=0, drop=True)
)
```

```python
shares_df.to_csv(r'../../Data/quotes_data/Shares.csv', index=False, encoding = 'utf-16', sep = '\t')
```

```python
bonds_df = pd.read_csv(r'../../Data/quotes_data/Bonds.csv', encoding = 'utf-16', sep = '\t')
```

```python
# ensure sorted by ISIN and date
bonds_df = bonds_df.sort_values(['SECID', 'TRADEDATE'])

# compute daily returns
bonds_df['ret'] = bonds_df.groupby('SECID')['LEGALCLOSEPRICE'].pct_change()

# 30-day rolling volatility of returns
bonds_df['ret_volatility'] = (
    bonds_df
    .groupby('SECID')['ret']
    .rolling(30, min_periods=30)
    .std()
    .reset_index(level=0, drop=True)
)

bonds_df['yield_volatility'] = (
    bonds_df
    .groupby('SECID')['YIELDCLOSE']
    .ffill()
    .rolling(30, min_periods=30)
    .std()
    .reset_index(level=0, drop=True)
)
```

```python
bonds_df.to_csv(r'../../Data/quotes_data/Bonds.csv', index=False, encoding = 'utf-16', sep = '\t')
```

```python
All_instr_df = pd.concat([shares_df, bonds_df], ignore_index=True)
```

```python
All_instr_df.to_csv(r'../../Data/quotes_data/All_instr_df.csv', index=False, encoding = 'utf-16', sep = '\t')
```
